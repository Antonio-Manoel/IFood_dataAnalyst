```python
df = df.dropna()
```

```python
df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"], format="%d-%m-%Y")
```

```python
df.info()
```

```python
df["Dt_Customer"].describe()
```

```python
df["DaysSinceEnrolled"] = df["Dt_Customer"].max() - df["Dt_Customer"]
df["DaysSinceEnrolled"] = df["DaysSinceEnrolled"].dt.days

df.head()
```

```python
df["YearsSinceEnrolled"] = df["DaysSinceEnrolled"] // 365
```

```python
df["Age"] = df["Dt_Customer"].max().year - df["Year_Birth"]
```

```python
# grupos de idade
# adulto_jovem: 18-30
# adulto: 31-45
# adulto_senior: 46-60
# idoso: 61-

df["AgeGroup"] = pd.cut(df["Age"], bins=[18, 30, 45, 60, df["Age"].max()],
                        labels=["18-30", "31-45", "46-60", "61+"], include_lowest=True)
```

```python
df["MntTotal"] = (df["MntWines"] + df["MntFruits"] + df["MntMeatProducts"] + df["MntFishProducts"] + df["MntSweetProducts"]
                  + df["MntGoldProds"])

df["MntRegularProds"] = df["MntTotal"] - df["MntGoldProds"]
```

```python
df["Children"] = df["Kidhome"] + df["Teenhome"]
```

```python
df["HasChildren"] = df["Children"].apply(lambda x: 1 if x > 0 else 0)
```

```python
df["Marital_Status"] = df["Marital_Status"].replace({
    "Alone": "Single",
    "Absurd": "Single",
    "YOLO": "Single",
    "Together": "Partner",
    "Married": "Partner",
    "Widow": "Single",
    "Single": "Single",
    "Divorced": "Single"
})
```

```python
df["AcceptedCmpTotal"] = (df["AcceptedCmp1"] + df["AcceptedCmp2"] + df["AcceptedCmp3"] + df["AcceptedCmp4"] + df["AcceptedCmp5"])

df["HasAcceptedCmp"] = df["AcceptedCmpTotal"].apply(lambda x: 1 if x > 0 else 0)
```
